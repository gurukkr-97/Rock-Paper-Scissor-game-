## ROCK PAPER SCISSOR

```python
import random
```

```python
print("*"*10, "WELCOME TO ROCK PAPER SCISSOR","*"*10)
name = input("Enter the Player name: ")
rnds = int(input("How many Rounds you want to Play"))
ch = ["rock","paper","scissor"]
sys_cnt = 0
user_cnt = 0
for i in range (1, rnds+1):
    sys_ch = random.choice(ch)
    if i == rnds:
        print("*****FINAL ROUND*****")
    else :
        print(f"*****ROUND {rnds}*****")
    user = input ("""Enter your Choice: 
                     *Rock
                     *Paper
                     *Scissor>>""")
    if user.lower() in ch:
        if user.lower() == sys_ch:
            print("ROUND DRAW")
            sys_cnt += 0.5
            user_cnt += 0.5
        elif ((user.lower() == "rock" and sys_ch == "scissor") or (user.lower() == "scissor" and sys_ch == "paper") or 
              (user.lower() == "paper" and sys_ch == "rock")):
            print(f"{name.upper()} WINS")
            user_cnt += 1
        else:
            print("COMPUTER WINS")
            sys_cnt += 1
    else:
        print("Invalid Entry, Please enter the valid input")
if user_cnt == sys_cnt:
    print(f"""
                SCORE BOARD:
                {name.upper()} : {user_cnt}
                COMPUTER : {sys_cnt}
                !!!!MATCH DRAW!!!!""")
elif user_cnt > sys_cnt:
    print(f"""
                SCORE BOARD:
                {name.upper()} : {user_cnt}
                COMPUTER : {sys_cnt}
                !!!!{name.upper()} WINS!!!!""")
else:
     print(f"""
                SCORE BOARD:
                {name.upper()} : {user_cnt}
                COMPUTER : {sys_cnt}
                !!!!COMPUTER WINS!!!!""")
```
